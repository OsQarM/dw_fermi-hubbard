```python
chain_length = 11

t = 1
U = 0
e = U
T = 4.5
dt = 0.01


nsteps = int(T/dt) -1
model_qutip = StandardFermiHubbard(chain_length, t, U, e, T, nsteps)

model_qutip.place_single_excitation('up', 0)
# model_qutip.place_single_excitation('up', 2)
# model_qutip.place_single_excitation('up', 4)
# model_qutip.place_single_excitation('up', 6)
# model_qutip.place_single_excitation('up', 8)
# model_qutip.place_single_excitation('down', 1)
# model_qutip.place_single_excitation('down', 3)
# model_qutip.place_single_excitation('down', 5)
model_qutip.place_single_excitation('down', chain_length-1)
pauli_strings = ['Z']
result, measurements = model_qutip.run(pauli_strings=pauli_strings)
```

```python
anim, fig = plot_chain_animation(measurements['Z'])

plt.close(fig)
HTML(anim.to_jshtml())
```
